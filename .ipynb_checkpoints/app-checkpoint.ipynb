```python
import gradio as gr
import cv2
import numpy as np
import joblib
import os

# Load model
model_path = "model/random_forest.pkl"

if not os.path.exists(model_path):
    raise Exception("Model not found! Train model first.")

model = joblib.load(model_path)

# Your categories
categories = ["Blot-Gel", "Histopathology", "X-Ray"]

# Prediction function
def predict_image(image):
    img = cv2.resize(image, (64, 64))
    img = img / 255.0
    img = img.flatten().reshape(1, -1)

    pred = model.predict(img)[0]
    label = categories[pred]

    return label

# Gradio UI
gr.Interface(
    fn=predict_image,
    inputs=gr.Image(type="numpy", label="Upload Image"),
    outputs=gr.Textbox(label="Prediction"),
    examples=[
        "samples/blot1.jpg",
        "samples/histo1.jpg",
        "samples/xray1.jpg"
    ]
).launch()
```
